```python
import pandas as pd
import plotly.express as px


# Dataset: Gapminder — GDP, Life Expectancy, Population by Country
# Source: Gapminder Foundation (gapminder.org)

df = px.data.gapminder()
print(f"Loaded: {len(df)} rows")
print(df.head())
```

```python
# explore

print(df.info())
print("Years:", sorted(df['year'].unique()))
print("Continents:", df['continent'].unique())
print(df.describe().round(1))
```

```python

# Task 1 — Scatter chart

# Select one continent and the two required years
continent_choice = "Asia"

df_task1 = df.loc[
    (df["continent"] == continent_choice) &
    (df["year"].isin([2002, 2007]))
].copy()

# Create the scatter chart
fig = px.scatter(
    df_task1,
    x="gdpPercap",
    y="lifeExp",
    color="year",
    color_discrete_map={
        2002: "#9CA3AF",
        2007: "#2563EB"
    },
    hover_name="country",
    log_x=True,
    title="Asia: GDP Growth and Life Expectancy Changes (2002–2007)",
    labels={
        "gdpPercap": "GDP per capita (US$, log scale)",
        "lifeExp": "Life expectancy (years)",
        "year": "Year"
    }
)

# Improve marker appearance
fig.update_traces(
    marker=dict(size=10, opacity=0.8)
)

# Clean layout
fig.update_layout(
    template="plotly_white",
    legend_title_text="Year"
)

fig.show()
```

```python

# Task 2 — Bubble chart

# Select all countries for 2007
df_task2 = df.loc[df["year"] == 2007].copy()

# Highlight Asia and show all other continents in grey
df_task2["Group"] = df_task2["continent"].apply(
    lambda continent: "Asia" if continent == "Asia"
    else "Other continents"
)

# Create the bubble chart
fig = px.scatter(
    df_task2,
    x="gdpPercap",
    y="lifeExp",
    size="pop",
    color="Group",
    color_discrete_map={
        "Asia": "#2563EB",
        "Other continents": "#D1D5DB"
    },
    hover_name="country",
    size_max=45,
    log_x=True,
    opacity=0.8,
    title="Asia Shows Wide Differences in Income and Life Expectancy (2007)",
    labels={
        "gdpPercap": "GDP per capita (US$, log scale)",
        "lifeExp": "Life expectancy (years)",
        "pop": "Population",
        "Group": "Region"
    }
)

# Add an annotation
fig.add_annotation(
    x=2500,
    y=73,
    text="Asia: diverse income and<br>life expectancy levels",
    showarrow=True,
    arrowhead=2,
    ax=80,
    ay=-55,
    bgcolor="white",
    bordercolor="#2563EB",
    borderwidth=1
)

# Clean layout
fig.update_layout(
    template="plotly_white",
    xaxis_type="log"
)

fig.show()
```
